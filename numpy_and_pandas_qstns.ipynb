```python
import pandas as pd
import numpy as np
```

```python

url = "https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv"

df = pd.read_csv(url)

```

```python
df=df.drop_duplicates()
print(df.duplicated().sum())
```

```python
df=df.rename(columns={"Title":"Movie_Title"})
print(df.columns)
```

```python
genres=df["Genre"].str.split(",")#split will help in seperate each dataset by commas
uni_genres=set(
    genre.strip()#set will help in store unique value ,if duplicate occurs it will not store it and strip removes extra spaces
    for genre_list in genres.dropna()#dropna skips missing values
    for genre in genre_list

)
print(uni_genres)
```

```python
df["Genre_count"]=df["Genre"].str.split(",").str.len()
df[["Movie_Title","Genre","Genre_count"]].head()
```

```python

ren_median=df["Revenue (Millions)"].median()
df["Revenue (Millions)"]=df["Revenue (Millions)"].fillna(ren_median)
print(ren_median)
```

```python
df[["Revenue (Millions)"]].isnull().sum()
```

```python
met_median=df["Metascore"].median()
df["Metascore"]=df["Metascore"].fillna(met_median)
print(met_median)
```

```python
df[["Metascore"]].isnull().sum()
```

```python
top_movies=df.sort_values(by="Rating",ascending=False).head()
print(top_movies[["Movie_Title","Rating"]])
```

```python
high_rev=df.loc[df["Revenue (Millions)"].idxmax()]#loc is uses that index to get the complte movie row
low_rev=df.loc[df["Revenue (Millions)"].idxmin()]

print("highest revenue movie:")
print(high_rev[["Movie_Title","Revenue (Millions)"]])
print("")
print("Lowest revenue movie:")
print(low_rev[["Movie_Title","Revenue (Millions)"]])
```

```python
avg_rate=df.groupby("Director")["Rating"].mean()
print(avg_rate)
```

```python
mov_yr=df.groupby("Year")["Movie_Title"].count()
print(mov_yr)
```

```python
sel_mov=df[(df["Rating"]>8)&(df["Revenue (Millions)"]>100)]
print(sel_mov[["Movie_Title","Rating","Revenue (Millions)"]])
```

```python
dir_sumry=df.groupby("Director").agg(
    avg_Rat=("Rating", "mean"),
    avg_Rev=("Revenue (Millions)","mean"))
print(dir_sumry)
```

```python
year_sumry=df.groupby("Year").agg(
    movie_Count=("Movie_Title", "count"),
    avg_Rat=("Rating", "mean")
)

print(year_sumry)
```

```python
df["Profitability"]=(
    df["Revenue (Millions)"]/df["Runtime (Minutes)"])

print(df[["Movie_Title","Revenue (Millions)","Runtime (Minutes)","Profitability"]].head())

print("")
```

```python
top_10_prof=df.sort_values(
    by="Profitability",ascending=False).head(10)

print(top_10_prof[
    ["Movie_Title","Profitability"]])
```

```python
def rating_catgry(rating):

    if rating<6:
        return "Poor"
    elif rating<7.5:
        return "Average"
    else:
        return "Excellent"


df["Rating_Category"]=df["Rating"].apply(rating_catgry)

print(df[["Movie_Title","Rating","Rating_Category"]].head())
```

```python
def movie_scre(row):
    return row["Rating"]*10+row["Revenue (Millions)"]

df["High_Score"]=df.apply(movie_scre,axis=1)

print(df[["Movie_Title","Rating","Revenue (Millions)","High_Score"]].head())
```
